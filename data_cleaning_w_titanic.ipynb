### NaN 값이 있는 행 제거하기 (결측치 제거)

```python
len(titanic_df)
```

```python
# 방법 1. 노이즈가 끼는 것을 막기 위해 결측치가 있는 모든 행 제거
titanic_df.dropna()

# 데이터의 크기가 충분히 크지 않으면, 정보의 유실이 커지기 때문에 충분히 고려해야 함
```

```python
# 방법 2. 컬럼을 지정하고 싶을 때
# 하나의 컬럼만 사용하더라도 리스트 형태로 정의해야 함
titanic_df.dropna(subset = ['Embarked'])
```


데이터가 많지 않은 경우에 결측치를 모두 제거하는 것은 좋지 않기 때문에,
해당 예제에서는 결측치를 채워줄 것이다.



```python
titanic_df['Age'].mean()
```

```python
titanic_df['Age'].median()
```

```python
# 일반적으로 median을 사용한다.
# mean의 경우, outlier 때문에 값이 치우칠 여지가 있다.

titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].median())
```

```python
titanic_df.loc[age_na_index]
```

```python
# 카테고리형 데이터의 결측치는 평균을 구할 수 없다.

# 방법 1. 특정 값이 다수의 역할을 하고 있을 때 (99% 값이 S인 경우)
titanic_df['Embarked'].value_counts()
```

```python
640 / 885
# 해당 예제에서는 약 72%로 방법 1이 최선은 아닐 수 있다.
```

```python
# 방법 2. null값임을 정의해주는 별도의 값으로 채우기
# unknown, N/A 등

titanic_df['Embarked'].fillna('N/A')
```

```python
# 해당 예제에서는 null값을 따로 정의해두는 게 큰 의미는 없을 것 같다.
# 따라서 72%의 확률을 믿고 채우기

titanic_df['Embarked'] = titanic_df['Embarked'].fillna('S')
```

특정 컬럼에 결측치가 많은 경우(50%, 70% 이상 등) 컬럼 자체를 제거하는 것도 방법이다.
